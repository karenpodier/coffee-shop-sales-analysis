### 1. Introduction and Menu Setup
In this section, we'll set up our coffee shop's menu using a Python dictionary. Dictionaries are great for storing key-value pairs, like item names and their prices.
#### Demo Example: Creating a Simple Menu

```python
# Our coffee shop menu
menu = {
    'Coffee': 2.50,
    'Tea': 2.00,
    'Croissant': 3.00,
    'Muffin': 2.75
}

print("Welcome to our Coffee Shop! Here's our menu:")
for item, price in menu.items():
    print(f"- {item}: £{price:.2f}")
```

### 2. Taking Orders
Now that we have our menu, let's allow customers to place orders. We'll use a `while` loop to let them order multiple items until they are done, and store their selections in a list.
#### Demo Example: Simple Order Taking

```python
# Initialize an empty list to store customer orders
customer_order = []

while True:
    order_item = input("What would you like to order? (Type 'done' to finish): ").strip().title()
    if order_item == 'Done':
        break
    elif order_item in menu:
        customer_order.append(order_item)
        print(f"Added {order_item} to your order.")
    else:
        print("Sorry, that item is not on our menu. Please choose from the list.")

print("Your current order: ", customer_order)
```

### 3. Calculating Bill
Now we'll calculate the total cost of the customer's order. This involves iterating through the `customer_order` list, looking up prices from the `menu` dictionary, and summing them up. We'll also add tax (VAT) to the total.
#### Demo Example: Calculating Subtotal and Total

```python
# Assuming customer_order is a list of items (from previous section's demo)
# If you completed the Learner Activity for section 2, adjust this to handle (item, quantity) tuples
customer_order_demo = ['Coffee', 'Muffin', 'Coffee'] # Example if not using quantity
# If using quantity, uncomment and use this example:
# customer_order_demo = [('Coffee', 2), ('Muffin', 1)]

subtotal = 0
for item in customer_order_demo:
    # Adjust this line if you are using (item, quantity) tuples
    subtotal += menu[item]

VAT_RATE = 0.20 # 20% VAT
tax_amount = subtotal * VAT_RATE
total_bill = subtotal + tax_amount

print(f"Subtotal: £{subtotal:.2f}")
print(f"VAT ({VAT_RATE*100:.0f}%): £{tax_amount:.2f}")
print(f"Total Bill: £{total_bill:.2f}")
```

### 4. Discounts and Lucky Draw
Let's add some fun! We'll implement a discount logic using `if-elif-else` statements and a 'lucky draw' feature using the `random` module.
#### Demo Example: Applying a Discount and Lucky Draw

```python
import random

# Assume total_bill is calculated from the previous section
total_bill_demo = 12.50 # Example total bill
discount_applied = 0

if total_bill_demo > 10:
    discount_percentage = 0.10 # 10% discount
    discount_amount = total_bill_demo * discount_percentage
    total_bill_demo -= discount_amount
    discount_applied = discount_amount
    print(f"Congratulations! You received a {discount_percentage*100:.0f}% discount of £{discount_amount:.2f}.")

print(f"New Total after discount: £{total_bill_demo:.2f}")

# Lucky Draw
lucky_prizes = ["free cookie", "free coffee", "10% off next order"]
if random.random() < 0.3: # 30% chance of winning
    prize = random.choice(lucky_prizes)
    print(f"*** Lucky Draw! You won a {prize}! ***")
else:
    print("Better luck next time in our Lucky Draw!")
```
